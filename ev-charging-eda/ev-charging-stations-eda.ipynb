## Imports, plotting config, and helper functions
 

```python
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path

import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display, Markdown

# Plot style
plt.rcParams["figure.figsize"] = (11, 5.5)
plt.rcParams["axes.spines.top"] = False
plt.rcParams["axes.spines.right"] = False
sns.set_style("whitegrid")

SEED = 42
rng = np.random.default_rng(SEED)

def section(title, subtitle=None):
    """Render a markdown section header from inside a code cell."""
    text = f"## {title}"
    if subtitle:
        text += f"\n\n{subtitle}"
    display(Markdown(text))

def hr():
    """Horizontal rule in markdown."""
    display(Markdown("---"))

def pct(x):
    """Format a ratio as a percentage string."""
    return f"{x * 100:.1f}%"
```

#  Load Data

```python
KAGGLE_DATA_DIR = Path("/kaggle/input/global-ev-charging-stations")
LOCAL_DATA_DIR = Path("/mnt/data/evdata")

DATA_DIR = KAGGLE_DATA_DIR if KAGGLE_DATA_DIR.exists() else LOCAL_DATA_DIR

if not DATA_DIR.exists():
    raise FileNotFoundError(
        "Dataset directory not found. Update DATA_DIR to the folder containing the CSV files."
    )

charging = pd.read_csv(DATA_DIR / "charging_station.csv")
country = pd.read_csv(DATA_DIR / "country_summary.csv")
ev_models = pd.read_csv(DATA_DIR / "ev_models.csv")
world = pd.read_csv(DATA_DIR / "world_summary.csv")
city = pd.read_csv(DATA_DIR / "charging_station_ml.csv")

section("Dataset overview", "First look at all tables: stations, cities, countries, and EV models.")
print(f"Data directory: {DATA_DIR}")
```

```python
section("Global snapshot", "High-level KPIs before diving into plots.")

POWER_MAX_REASONABLE_KW = 1000
MIN_STATIONS_FOR_RANKING = 100

countries_covered = charging["country_code"].nunique()
cities_covered = charging["city"].nunique()
fast_share_global = charging["is_fast_dc"].mean()

raw_power = pd.to_numeric(charging["power_kw"], errors="coerce")
highest_reported_power = raw_power.max()
p99_power = raw_power.quantile(0.99)
extreme_power_rows = (raw_power > POWER_MAX_REASONABLE_KW).sum()

top_make_counts = ev_models["make"].value_counts()
top_make = top_make_counts.idxmax() if len(top_make_counts) else None
top_make_n = top_make_counts.max() if len(top_make_counts) else 0

kpi = pd.DataFrame(
    {
        "Metric": [
            "Total stations",
            "Countries with stations",
            "Cities with station labels",
            "Global fast DC share",
            "Highest reported raw power",
            "99th percentile site power (kW)",
            "Extreme reported power rows (>1000 kW)",
            "Top EV make by number of models",
        ],
        "Value": [
            f"{len(charging):,}",
            f"{countries_covered}",
            f"{cities_covered}",
            pct(fast_share_global),
            f"{highest_reported_power:,.0f} kW (retained; excluded from robust aggregates)",
            f"{p99_power:,.0f}",
            f"{extreme_power_rows:,}",
            f"{top_make} ({top_make_n})" if top_make is not None else "N/A",
        ],
    }
)

display(kpi)
hr()
```

## Validation notes & analysis safeguards



```python
section(
    "Validation notes and analysis safeguards",
    "Reported value coverage, duplicates, extreme records, and robust analysis-safe columns."
)

print("Missing values per table:\n")
for df, name in [
    (charging, "Charging"),
    (country, "Country"),
    (city, "City"),
    (ev_models, "EV Models"),
    (world, "World"),
]:
    print(f"{name}:")
    display(df.isna().sum())
    print()

print("Duplicate rows:")
print("Charging:", charging.duplicated().sum())
print("Country:", country.duplicated().sum())
print("City:", city.duplicated().sum())
print("EV Models:", ev_models.duplicated().sum())
print("World:", world.duplicated().sum())

# ---- Conservative preparation ----

# Normalize city/state text for missing values while preserving location-placeholder buckets.
charging["city"] = charging["city"].fillna("Unknown City")
charging["state_province"] = charging["state_province"].fillna("UNKNOWN")

# Ensure country_code exists for stations and countries.
charging = charging.dropna(subset=["country_code"]).copy()
country = country.dropna(subset=["country_code"]).copy()

# Ensure numeric types without overwriting raw fields with imputed values.
charging["power_kw"] = pd.to_numeric(charging["power_kw"], errors="coerce")
charging["ports"] = pd.to_numeric(charging["ports"], errors="coerce")

# Coordinate range validation.
before_geo = len(charging)
charging = charging[
    charging["latitude"].between(-90, 90)
    & charging["longitude"].between(-180, 180)
].copy()
after_geo = len(charging)
print(f"\nRows outside valid coordinate ranges removed: {before_geo - after_geo:,}")

# Validation summary after basic schema and coordinate checks.
validation_summary = pd.DataFrame(
    {
        "Validation note": [
            "Missing power_kw metadata",
            f"Extreme reported power_kw > {POWER_MAX_REASONABLE_KW}",
            "Non-positive port-count rows",
            "Large port-count rows > 50",
            "Placeholder city labels",
            "Placeholder state/province labels",
        ],
        "Rows": [
            charging["power_kw"].isna().sum(),
            (charging["power_kw"] > POWER_MAX_REASONABLE_KW).sum(),
            (charging["ports"] <= 0).sum(),
            (charging["ports"] > 50).sum(),
            charging["city"].eq("Unknown City").sum(),
            charging["state_province"].eq("UNKNOWN").sum(),
        ],
    }
)

validation_summary["Share"] = validation_summary["Rows"] / len(charging)
validation_summary["Share"] = validation_summary["Share"].map(lambda x: f"{x:.2%}")

print("\nValidation notes used to guide robust aggregate metrics:")
display(validation_summary)

print(
    "\nAnalysis safeguard: raw records are preserved, while robust aggregate metrics use "
    "analysis-safe columns where extreme reported values could distort rankings or averages."
)

extreme_power_examples = (
    charging
    .loc[charging["power_kw"] > POWER_MAX_REASONABLE_KW,
         ["id", "name", "country_code", "city", "ports", "power_kw", "power_class", "is_fast_dc"]]
    .sort_values("power_kw", ascending=False)
)

if len(extreme_power_examples):
    print("\nHighest reported power records retained in raw data and excluded from robust power aggregates:")
    display(extreme_power_examples.head(10))

# Preserve raw columns and create analysis-safe columns.
charging["power_kw_clean"] = charging["power_kw"].where(
    charging["power_kw"].between(0, POWER_MAX_REASONABLE_KW)
)
power_median = charging["power_kw_clean"].median()
charging["power_kw_filled"] = charging["power_kw_clean"].fillna(power_median)

charging["ports_clean"] = charging["ports"].where(charging["ports"] > 0)
ports_median = charging["ports_clean"].median()
charging["ports_filled"] = charging["ports_clean"].fillna(ports_median)

# Cast to appropriate dtypes.
charging["country_code"] = charging["country_code"].astype("category")
if "power_class" in charging.columns:
    charging["power_class"] = charging["power_class"].astype("category")

charging["is_fast_dc"] = charging["is_fast_dc"].astype(bool)

print("\nFinal shapes after preparation:")
print("Charging:", charging.shape)
print("Country:", country.shape)
print("City:", city.shape)
print("World:", world.shape)
print("EV Models:", ev_models.shape)

hr()
```

## Feature engineering & country-level dashboard

```python
section(
    "Country-level infrastructure profile",
    "Aggregate station-level data into a country dashboard using analysis-safe power and port fields."
)

# Station type label
charging["station_type"] = charging["is_fast_dc"].map(
    {True: "Fast DC", False: "Normal / AC"}
)

# Power buckets for nicer plots using analysis-safe power values.
power_bins = [0, 22, 50, 150, 350, np.inf]
power_labels = ["≤22kW AC", "22–50kW", "50–150kW", "150–350kW", ">350kW"]

power_kw_viz = charging["power_kw_filled"].clip(
    upper=charging["power_kw_filled"].quantile(0.995)
)
charging["power_bucket"] = pd.cut(
    power_kw_viz, bins=power_bins, labels=power_labels
)

# Country-level dashboard from charging table.
country_dash = (
    charging.groupby("country_code", observed=False)
    .agg(
        stations=("id", "count"),
        ports=("ports_clean", "sum"),
        avg_ports=("ports_clean", "mean"),
        fast_share=("is_fast_dc", "mean"),
        p99_power=("power_kw_clean", lambda s: s.quantile(0.99)),
    )
    .reset_index()
)

country_dash["fast_share_pct"] = country_dash["fast_share"] * 100
country_dash["ports_per_station"] = country_dash["ports"] / country_dash["stations"]

# Top known city per country by station count.
known_city_rows = charging[charging["city"].str.lower().ne("unknown city")].copy()

top_city = (
    known_city_rows.groupby(["country_code", "city"], observed=False)
    .size()
    .reset_index(name="stations_city")
    .sort_values(["country_code", "stations_city"], ascending=[True, False])
    .drop_duplicates("country_code")
)

country_dash = country_dash.merge(
    top_city[["country_code", "city", "stations_city"]],
    on="country_code",
    how="left",
)

display(country_dash.sort_values("stations", ascending=False).head(10))

hr()
```

## Country typologies: coverage vs fast DC share

```python
section(
    "Country typologies",
    "Classifying countries by coverage (stations) and fast DC share."
)

df_typo = country_dash.copy()

# Thresholds: median-based (can be tuned)
stations_thr = df_typo["stations"].median()
fast_share_thr = df_typo["fast_share"].median()

def coverage_level(x):
    return "High coverage" if x >= stations_thr else "Low coverage"

def fast_level(x):
    return "High fast share" if x >= fast_share_thr else "Low fast share"

df_typo["coverage"] = df_typo["stations"].apply(coverage_level)
df_typo["fast_level"] = df_typo["fast_share"].apply(fast_level)
df_typo["typology"] = df_typo["coverage"] + " / " + df_typo["fast_level"]

display(
    df_typo.groupby("typology")[["country_code"]].count().rename(columns={"country_code": "n_countries"})
)

plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df_typo,
    x="stations",
    y="fast_share",
    hue="typology",
    style="typology",
    alpha=0.8,
)
plt.xscale("log")
plt.xlabel("Stations (log scale)")
plt.ylabel("Fast DC share")
plt.title("Country typology: coverage vs fast DC share")
plt.tight_layout()
plt.show()

hr()
```

## Consistency checks between station, city, and country tables

```python
section(
    "Consistency checks",
    "Do aggregated counts from stations roughly match city and country tables?"
)

# 1) Stations → Country (from raw charging)
stations_by_country_from_raw = (
    charging.groupby("country_code")["id"].count().rename("stations_raw")
)

# 2) Stations → Country (from city table)
stations_by_country_from_city = (
    city.groupby("country_code")["station_count"].sum().rename("stations_city_sum")
)

consistency = (
    pd.concat([stations_by_country_from_raw, stations_by_country_from_city], axis=1)
    .dropna()
)
consistency["diff"] = consistency["stations_raw"] - consistency["stations_city_sum"]
consistency["rel_diff_pct"] = (
    consistency["diff"].abs() / consistency["stations_raw"].clip(lower=1) * 100
)

print("Consistency summary between raw stations and city-level station_count:")
display(consistency.describe()[["diff", "rel_diff_pct"]])

# Show largest relative differences
display(
    consistency.sort_values("rel_diff_pct", ascending=False).head(10)
)

hr()
```

## Leakage-aware baseline — City fast-DC presence


```python
section(
    "Leakage-aware baseline — City fast-DC presence",
    "A transparent baseline using non-power city features only. Power-derived fields are excluded to avoid circular modeling."
)

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, accuracy_score
from sklearn.dummy import DummyClassifier

# ---- 0) Guard: make sure target exists ----
if "has_fast_dc" not in city.columns:
    print("Column `has_fast_dc` not found in city table — skipping baseline ML block.")
    hr()
else:
    # ---- 1) Basic dataset summary ----
    city_model = city_known.copy()
    target = city_model["has_fast_dc"].astype(int)
    pos_rate = target.mean()

    print(f"Known cities used for modeling: {len(city_model):,}")
    print(f"Share of known cities with at least one fast DC station: {pct(pos_rate)}")

    # ---- 2) Feature-definition check ----
    feature_definition_check = pd.crosstab(
        city_model["max_power_kw"].ge(50).astype(int),
        city_model["has_fast_dc"].astype(int),
        rownames=["max_power_kw >= 50"],
        colnames=["has_fast_dc"],
    )

    print(
        "\nFeature-definition check: power-derived fields are intentionally excluded "
        "because they are very close to the fast-DC label definition."
    )
    display(feature_definition_check)

    # ---- 3) Features & split ----
    # Exclude max_power_kw and median_power_kw because they are too close to the fast-DC label definition.
    features = ["station_count", "port_count"]
    X = city_model[features].fillna(0)
    y = target

    X_train, X_valid, y_train, y_valid = train_test_split(
        X, y, test_size=0.25, random_state=SEED, stratify=y
    )

    metrics = {}

    # ---- 4) Dummy baseline: always predict majority class ----
    dummy = DummyClassifier(strategy="most_frequent")
    dummy.fit(X_train, y_train)

    y_dummy = dummy.predict(X_valid)
    y_dummy_proba = np.full_like(y_dummy, y_train.mean(), dtype=float)

    metrics["Dummy (majority class)"] = {
        "ROC-AUC": roc_auc_score(y_valid, y_dummy_proba),
        "Accuracy": accuracy_score(y_valid, y_dummy),
    }

    # ---- 5) Baseline model: Logistic Regression ----
    clf = LogisticRegression(max_iter=1000, class_weight="balanced")
    clf.fit(X_train, y_train)

    y_pred_proba = clf.predict_proba(X_valid)[:, 1]
    y_pred = (y_pred_proba >= 0.5).astype(int)

    metrics["Logistic (non-power features)"] = {
        "ROC-AUC": roc_auc_score(y_valid, y_pred_proba),
        "Accuracy": accuracy_score(y_valid, y_pred),
    }

    # ---- 6) Compare models in a single table ----
    results_df = (
        pd.DataFrame(metrics)
        .T[["ROC-AUC", "Accuracy"]]
        .round(3)
        .sort_values("ROC-AUC", ascending=False)
    )

    print("\nValidation performance comparison:")
    display(results_df)

    # ---- 7) Coefficients as simple feature effects ----
    coef_df = (
        pd.DataFrame({"feature": features, "coef": clf.coef_[0]})
        .sort_values("coef", key=np.abs, ascending=False)
        .reset_index(drop=True)
    )

    print("\nFeature effects (log-odds coefficients for the positive class):")
    display(coef_df)

    hr()
```

## City clustering: infrastructure profiles

```python
section(
    "City clusters",
    "Grouping known cities into infrastructure profiles with KMeans using robust power features."
)

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

city_clust = city_known.copy()

city_clust["max_power_kw_clean"] = city_clust["max_power_kw"].where(
    city_clust["max_power_kw"].between(0, POWER_MAX_REASONABLE_KW)
)
city_clust["median_power_kw_clean"] = city_clust["median_power_kw"].where(
    city_clust["median_power_kw"].between(0, POWER_MAX_REASONABLE_KW)
)

cluster_features = [
    "station_count",
    "port_count",
    "fast_station_share",
    "fast_port_share",
    "max_power_kw_clean",
    "median_power_kw_clean",
]

X = city_clust[cluster_features].fillna(0)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

kmeans = KMeans(n_clusters=4, random_state=SEED, n_init=10)
city_clust["cluster"] = kmeans.fit_predict(X_scaled)

# Cluster-level summary
cluster_summary = (
    city_clust
    .groupby("cluster")[cluster_features]
    .mean()
    .round(2)
    .sort_index()
)

display(cluster_summary)

plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=city_clust,
    x="station_count",
    y="fast_station_share",
    hue="cluster",
    alpha=0.6,
)
plt.xscale("log")
plt.xlabel("Station count (log scale)")
plt.ylabel("Fast DC share")
plt.title("City clusters by station count and fast DC share")
plt.tight_layout()
plt.show()

hr()
```
